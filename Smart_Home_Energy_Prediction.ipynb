```python

# Step 1: Download the complete dataset

import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00374/energydata_complete.csv"

df = pd.read_csv(url)

print("Dataset downloaded successfully!")
print("Number of rows and columns:", df.shape)
```

```python

# Display the first 5 records
display(df.head())

# Display all column names
print("\nColumn names:")
print(df.columns.tolist())

# Check missing values
print("\nMissing values:", df.isnull().sum().sum())

# Check dataset size
print("\nDataset shape:", df.shape)
```

```python

# Step 2: Explore the dataset

# Summary of numerical columns
display(df.describe())

# Data types of all columns
print("\nColumn data types:")
print(df.dtypes)

# Check duplicate records
print("\nDuplicate rows:", df.duplicated().sum())

# Check the target variable
print("\nAppliance energy consumption:")
print(df["Appliances"].describe())
```

```python

# Step 3: Prepare features and target

# Remove date and random variables
df_ml = df.drop(columns=["date", "rv1", "rv2"])

# Separate input features and target
X = df_ml.drop(columns=["Appliances"])
y = df_ml["Appliances"]

print("Input features shape:", X.shape)
print("Target shape:", y.shape)

print("\nInput feature names:")
print(X.columns.tolist())

print("\nTarget variable:", y.name)
```

```python

from sklearn.model_selection import KFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.ensemble import AdaBoostRegressor
from sklearn.tree import DecisionTreeRegressor
import numpy as np
import pandas as pd
import time

# Define 10-fold cross-validation
cv = KFold(n_splits=10, shuffle=True, random_state=42)

# Define five algorithms
models = {
    "SVR": make_pipeline(StandardScaler(), SVR()),
    "Ridge Regression": make_pipeline(StandardScaler(), Ridge()),
    "Lasso Regression": make_pipeline(StandardScaler(), Lasso(max_iter=10000)),
    "Elastic Net": make_pipeline(StandardScaler(), ElasticNet(max_iter=10000)),
    "AdaBoost": AdaBoostRegressor(
        estimator=DecisionTreeRegressor(max_depth=4),
        n_estimators=50,
        random_state=42
    )
}

# Evaluate each algorithm
results = []

for name, model in models.items():
    start = time.time()

    scores = cross_val_score(
        model, X, y,
        cv=cv,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1
    )

    results.append({
        "Algorithm": name,
        "Mean RMSE": -scores.mean(),
        "Std RMSE": scores.std(),
        "Time (seconds)": round(time.time() - start, 2)
    })

# Display results
results_df = pd.DataFrame(results).sort_values("Mean RMSE")
display(results_df.round(2))
```

```python

import pickle
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge

# Train the best model on the full dataset
best_model = make_pipeline(
    StandardScaler(),
    Ridge()
)

best_model.fit(X, y)

# Save the trained model
with open("energy_model.pkl", "wb") as file:
    pickle.dump(best_model, file)

print("Model trained and saved successfully!")
```
